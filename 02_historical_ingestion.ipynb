### Download Historical Backfill Dataset 
* kaggle f1 dataset (1950-2024)

```python
import os 

os.environ["KAGGLE_USERNAME"]=dbutils.secrets.get(
    catalog="f1_warehouse",
    schema="ingestion",
    key="kaggle_username"
)

os.environ["KAGGLE_KEY"]=dbutils.secrets.get(
    catalog="f1_warehouse",
    schema="ingestion",
    key="kaggle_key"
)
```

```python
#use kaggle api to download files directly into volumes unlike using kagglehub which wld lead to less control over downloaded location

%pip install kaggle
from kaggle.api.kaggle_api_extended import KaggleApi
api = KaggleApi()
api.authenticate()
api.dataset_download_files( "rohanrao/formula-1-world-championship-1950-2020",
    path="/Volumes/f1_warehouse/ingestion/raw_files/kaggle",
    unzip=True)
```

```python
#check if all files loaded into temp

data_path=sorted(os.listdir("/Volumes/f1_warehouse/ingestion/raw_files/kaggle"))

for file in data_path:
    print(file)
```
